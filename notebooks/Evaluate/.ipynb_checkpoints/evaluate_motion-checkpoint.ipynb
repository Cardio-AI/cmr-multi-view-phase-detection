```python
def get_files(path, search_str=None):
    """
    Returns all files in a directory given with path.
    Optional: Give a search str, than it returns only the files with the given search_str
    """
    if search_str:
        files = sorted(glob.glob(os.path.join(path, search_str)))
    else:
        files = sorted(glob.glob(path))
    print(len(files))
    return files

def get_df(_path, file_name='dataset_information.csv'):
    """
    Returns the dataframe with the given file_name in the directory _path.
    """
    dataset_info_path = get_file(os.path.join(_path, file_name))
    df = pd.read_csv(dataset_info_path[0])
    # Find the first row with NaN values
    index_of_nan = df.index[df.isnull().any(axis=1)].tolist()

    # Stop reading at the first row with NaN values
    if index_of_nan:
        df = df.iloc[:index_of_nan[0]]
    return df

def get_files(_path, regex):
    path_ = os.path.join(_path, regex)
    images = sorted(glob.glob(path_))
    print(f"Found {len(images)} files in {_path} with regex {regex}.")
```
